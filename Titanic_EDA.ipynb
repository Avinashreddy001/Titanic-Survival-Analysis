# Analysis

```python
# 1. What was the overall survival rate?
overall_survival_rate=df.Survived.mean()
print(f"Overall Survival Rate: {overall_survival_rate:.2f}")
```

```python
# Create a simple bar chart showing survival by gender
plt.figure(figsize=(8, 5))
df.groupby('Sex')['Survived'].mean().plot(kind='bar', color=['blue', 'pink'])
plt.title('How Many Men vs Women Survived?')
plt.ylabel('Survival Rate (0 to 1)')
plt.xlabel('Gender')
plt.xticks(rotation=0)
plt.show()
```

```python
# 5. Survival Rate by Class and Gender
plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Pclass', y='Survived', hue='Sex')
plt.title('Survival Rate by Class and Gender')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.show()
```

```python
# Line chart - Survival by Age Groups
plt.figure(figsize=(10, 6))
age_survival = df.groupby('Age')['Survived'].mean().plot(kind='line')
plt.title('Survival Rate by Age')
plt.xlabel('Age')
plt.ylabel('Survival Rate')
plt.ylim(0,1)
plt.grid(True)
plt.show()
```

```python
# Histogram - Age Distribution
plt.figure(figsize=(10, 6))
plt.hist(df['Age'],
         bins=20,  # Divide data into 20 groups
         color='skyblue',
         edgecolor='black')  # Black outline on bars
plt.title('Age Distribution of Passengers', fontsize=14, fontweight='bold')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.axvline(df['Age'].mean(), color='red', linestyle='--', label=f'Average: {df["Age"].mean():.1f}')  # Red line at average
plt.legend()
plt.show()
```

```python
# Box plot - Age by Survival
plt.figure(figsize=(8, 6))

df.boxplot(column='Age', by='Survived',
           figsize=(8, 6),
           patch_artist=True)

plt.title('Age Distribution: Survivors vs Non-Survivors', fontsize=14, fontweight='bold')
plt.suptitle('')  # Remove automatic title
plt.xlabel('Survived (0=No, 1=Yes)')
plt.ylabel('Age')
plt.show()
```

```python
# Heatmap - Correlation between features
plt.figure(figsize=(10, 8))

# Calculate correlations (how things relate to each other)
numeric_df = df[['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']]
correlation = numeric_df.corr()

# Create heatmap
sns.heatmap(correlation,
            annot=True,  # Show numbers in cells
            cmap='coolwarm',  # Red=positive, Blue=negative
            center=0,  # White at zero
            square=True,  # Make cells square
            linewidths=1,  # Lines between cells
            fmt='.2f')  # Show 2 decimal places

plt.title('How Features Relate to Each Other', fontsize=14, fontweight='bold')
plt.show()
```

```python
# Scatter plot - Age vs Fare
plt.figure(figsize=(10, 6))

plt.scatter(df['Age'], df['Fare'],
            c=df['Survived'],  # Color by survival (0=red, 1=blue)
            cmap='coolwarm',
            alpha=0.6,  # Make dots slightly transparent
            s=50)  # Size of dots

plt.title('Age vs Fare Paid (colored by survival)', fontsize=14, fontweight='bold')
plt.xlabel('Age')
plt.ylabel('Fare Paid')
plt.colorbar(label='Survived')
plt.show()
```

```python
# Count plot - Passengers per class
plt.figure(figsize=(8, 6))

sns.countplot(data=df, x='Pclass', palette='Set2')

plt.title('Number of Passengers in Each Class', fontsize=14, fontweight='bold')
plt.xlabel('Passenger Class')
plt.ylabel('Count')

# Add count numbers on top of bars
for i, count in enumerate(df['Pclass'].value_counts().sort_index()):
    plt.text(i, count + 10, str(count), ha='center', fontweight='bold')

plt.show()
```

```python
# 6. Correlation Heatmap
plt.figure(figsize=(10,8))
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
corr_matrix = df[numeric_cols].corr()

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
# === VISUALIZATION 1: Survival by Gender ===

plt.figure(figsize=(10, 6))

# Calculate survival rate by gender
gender_survival = df.groupby('Sex')['Survived'].mean()

# Create bar chart
gender_survival.plot(kind='bar', color=['skyblue', 'lightcoral'])

# Add labels and title
plt.title('Survival Rate by Gender', fontsize=16, fontweight='bold')
plt.ylabel('Survival Rate', fontsize=12)
plt.xlabel('Gender', fontsize=12)
plt.xticks(rotation=0)  # Keep text horizontal
plt.ylim(0, 1)  # Y-axis from 0 to 1

# Add percentage labels on top of bars
for i, v in enumerate(gender_survival.values):
    plt.text(i, v + 0.02, f'{v*100:.1f}%', ha='center', fontweight='bold', fontsize=11)

plt.tight_layout()
plt.show()

# Print the numbers
print("Survival Rate by Gender:")
print(f"Female: {gender_survival['female']*100:.1f}%")
print(f"Male: {gender_survival['male']*100:.1f}%")
```

```python
# === VISUALIZATION 2: Overall Survival Distribution ===

plt.figure(figsize=(8, 8))

# Count survivors vs non-survivors
survival_counts = df['Survived'].value_counts()

# Create pie chart
colors = ['#ff6b6b', '#51cf66']  # Red for died, green for survived
explode = (0.05, 0)  # Slightly separate the "died" slice

plt.pie(survival_counts,
        labels=['Died', 'Survived'],
        autopct='%1.1f%%',  # Show percentages
        startangle=90,
        colors=colors,
        explode=explode,
        shadow=True,  # Add shadow for 3D effect
        textprops={'fontsize': 12, 'fontweight': 'bold'})

plt.title('Titanic Survival Distribution', fontsize=16, fontweight='bold', pad=20)
plt.show()

# Print the actual numbers
print("\nSurvival Statistics:")
print(f"Total Passengers: {len(df)}")
print(f"Survived: {survival_counts[1]} ({survival_counts[1]/len(df)*100:.1f}%)")
print(f"Died: {survival_counts[0]} ({survival_counts[0]/len(df)*100:.1f}%)")
```

```python
# === VISUALIZATION 3: Age Distribution ===

plt.figure(figsize=(12, 6))

# Create histogram with two colors for survived vs died
survived_ages = df[df['Survived'] == 1]['Age']
died_ages = df[df['Survived'] == 0]['Age']

plt.hist([died_ages, survived_ages],
         bins=20,  # 20 age groups
         label=['Died', 'Survived'],
         color=['#ff6b6b', '#51cf66'],
         alpha=0.7,  # Slightly transparent
         edgecolor='black')

# Add average age lines
avg_died = died_ages.mean()
avg_survived = survived_ages.mean()

plt.axvline(avg_died, color='red', linestyle='--', linewidth=2, label=f'Avg Died: {avg_died:.1f}')
plt.axvline(avg_survived, color='green', linestyle='--', linewidth=2, label=f'Avg Survived: {avg_survived:.1f}')

# Labels
plt.title('Age Distribution: Survivors vs Non-Survivors', fontsize=16, fontweight='bold')
plt.xlabel('Age', fontsize=12)
plt.ylabel('Number of Passengers', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Print statistics
print("\nAge Statistics:")
print(f"Average age of survivors: {avg_survived:.1f} years")
print(f"Average age of non-survivors: {avg_died:.1f} years")
print(f"Youngest passenger: {df['Age'].min():.1f} years")
print(f"Oldest passenger: {df['Age'].max():.1f} years")
```

## 📊 Key Visual Insights

### 1. Gender was the strongest survival factor
- **74.2%** of women survived vs only **18.9%** of men
- "Women and children first" policy was clearly followed
- Gender alone increased survival chances by **4x**

### 2. Overall survival was tragically low
- Only **38.4%** of passengers survived
- **549 people died** out of 891 total passengers
- This was one of the deadliest maritime disasters in history

### 3. Age played a minor role
- Survivors were slightly younger (28.3 years) vs non-survivors (30.0 years)
- However, the difference was minimal (only 1.7 years)
- Age distribution was similar for both groups

```python
# === VISUALIZATION 4: Survival by Passenger Class ===

plt.figure(figsize=(10, 6))

# Calculate survival rate by class
class_survival = df.groupby('Pclass')['Survived'].mean()

# Create bar chart with custom colors
colors_class = ['gold', 'silver', '#CD7F32']  # Gold, Silver, Bronze
class_survival.plot(kind='bar', color=colors_class, edgecolor='black', linewidth=1.5)

# Labels and title
plt.title('Survival Rate by Passenger Class', fontsize=16, fontweight='bold')
plt.xlabel('Passenger Class', fontsize=12)
plt.ylabel('Survival Rate', fontsize=12)
plt.xticks([0, 1, 2], ['1st Class\n(Upper)', '2nd Class\n(Middle)', '3rd Class\n(Lower)'], rotation=0)
plt.ylim(0, 1)

# Add percentage labels on bars
for i, v in enumerate(class_survival.values):
    plt.text(i, v + 0.02, f'{v*100:.1f}%', ha='center', fontweight='bold', fontsize=12)

plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# Print the statistics
print("Survival Rate by Passenger Class:")
for pclass, rate in class_survival.items():
    print(f"Class {pclass}: {rate*100:.1f}% ({df[df['Pclass']==pclass]['Survived'].sum()} survived out of {len(df[df['Pclass']==pclass])})")
```

```python
# === VISUALIZATION 5: Comprehensive Dashboard ===

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Titanic Survival Analysis Dashboard', fontsize=18, fontweight='bold', y=0.995)

# Panel 1: Survival by Gender
gender_data = df.groupby('Sex')['Survived'].mean()
axes[0, 0].bar(gender_data.index, gender_data.values, color=['#3498db', '#e74c3c'], edgecolor='black')
axes[0, 0].set_title('Survival Rate by Gender', fontsize=14, fontweight='bold')
axes[0, 0].set_ylabel('Survival Rate')
axes[0, 0].set_ylim(0, 1)
for i, v in enumerate(gender_data.values):
    axes[0, 0].text(i, v + 0.02, f'{v*100:.1f}%', ha='center', fontweight='bold')

# Panel 2: Survival by Class
class_data = df.groupby('Pclass')['Survived'].mean()
axes[0, 1].bar(['1st', '2nd', '3rd'], class_data.values, color=['#f39c12', '#95a5a6', '#d35400'], edgecolor='black')
axes[0, 1].set_title('Survival Rate by Class', fontsize=14, fontweight='bold')
axes[0, 1].set_ylabel('Survival Rate')
axes[0, 1].set_ylim(0, 1)
for i, v in enumerate(class_data.values):
    axes[0, 1].text(i, v + 0.02, f'{v*100:.1f}%', ha='center', fontweight='bold')

# Panel 3: Age Distribution
axes[1, 0].hist([df[df['Survived']==0]['Age'], df[df['Survived']==1]['Age']],
                bins=20, label=['Died', 'Survived'], color=['#e74c3c', '#2ecc71'], alpha=0.7)
axes[1, 0].set_title('Age Distribution by Survival', fontsize=14, fontweight='bold')
axes[1, 0].set_xlabel('Age')
axes[1, 0].set_ylabel('Count')
axes[1, 0].legend()
axes[1, 0].grid(alpha=0.3)

# Panel 4: Survival Count (Pie)
survival_counts = df['Survived'].value_counts()
axes[1, 1].pie(survival_counts, labels=['Died', 'Survived'], autopct='%1.1f%%',
               colors=['#e74c3c', '#2ecc71'], startangle=90, explode=(0.05, 0))
axes[1, 1].set_title('Overall Survival Distribution', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()

print("\n" + "="*60)
print("DASHBOARD SUMMARY")
print("="*60)
print(f"Total Passengers: {len(df)}")
print(f"Survived: {survival_counts[1]} ({survival_counts[1]/len(df)*100:.1f}%)")
print(f"Died: {survival_counts[0]} ({survival_counts[0]/len(df)*100:.1f}%)")
print(f"\nFemale Survival: {df[df['Sex']=='female']['Survived'].mean()*100:.1f}%")
print(f"Male Survival: {df[df['Sex']=='male']['Survived'].mean()*100:.1f}%")
print(f"\n1st Class Survival: {df[df['Pclass']==1]['Survived'].mean()*100:.1f}%")
print(f"3rd Class Survival: {df[df['Pclass']==3]['Survived'].mean()*100:.1f}%")
print("="*60)
```

## 🎯 Final Conclusions

### What the Data Tells Us:

**1. Social Class Determined Survival** 💰
- 1st class passengers had a **63%** survival rate
- 3rd class passengers had only a **24%** survival rate
- **Wealthier passengers had 2.6x better chance of survival**
- Location on ship and lifeboat access favored upper classes

**2. Gender Was the Strongest Factor** 👨👩
- **"Women and children first"** policy was strictly followed
- Women had **74%** survival rate vs men's **19%**
- Only **1 in 5 men survived** vs **3 in 4 women**

**3. Age Had Minimal Direct Impact** 👶👴
- Average age of survivors: **28.3 years**
- Average age of non-survivors: **30.0 years**
- Difference of only **1.7 years** (not statistically significant)
- However, children were prioritized within gender groups

**4. The Tragic Reality** ⚓
- **Only 38.4%** survived (342 out of 891)
- **549 people died** in this disaster
- Social inequality and gender norms directly impacted survival
